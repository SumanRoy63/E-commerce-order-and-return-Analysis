```python
# Phase 5: Dashboard (Streamlit)
from pathlib import Path

import pandas as pd
import streamlit as st
import seaborn as sns
import matplotlib.pyplot as plt

# Load cleaned dataset
workspace_root = Path.cwd()
data_path = workspace_root / "cleaned_ecommerce_returns.csv"
if not data_path.exists():
    data_path = workspace_root.parent / "cleaned_ecommerce_returns.csv"

df = pd.read_csv(data_path)

# --- Dashboard Title ---
st.title(" Return & Refund Pattern Analysis Dashboard")

# --- KPI Metrics ---
total_orders = len(df)
total_returns = df["is_return"].sum()
return_rate = round((total_returns / total_orders) * 100, 2)

# Convert relevant date columns and compute a return cycle using available data
df["delivered_date"] = pd.to_datetime(df["delivered_date"], errors="coerce")
df["request_date"] = pd.to_datetime(df["request_date"], errors="coerce")
df["return_cycle_days"] = (df["request_date"] - df["delivered_date"]).dt.days

avg_refund_cycle = (
    df.loc[df["returned"].eq("Yes"), "return_cycle_days"]
    .dropna()
    .mean()
)

st.metric("Total Orders", total_orders)
st.metric("Total Returns", total_returns)
st.metric("Return Rate (%)", return_rate)
st.metric("Avg Refund Cycle (days)", round(avg_refund_cycle, 2))

# --- Return Rate by Category ---
st.subheader("Return Rate by Category")
category_returns = df.groupby("category")["is_return"].mean().reset_index()

fig, ax = plt.subplots(figsize=(8,5))
sns.barplot(x="category", y="is_return", data=category_returns, palette="viridis", ax=ax)
ax.set_title("Return Rate by Category")
ax.set_ylabel("Return Rate (%)")
st.pyplot(fig)

# --- Refund Cycle Distribution ---
st.subheader("Refund Cycle Time Distribution")
return_cycle_data = df.loc[df["returned"].eq("Yes"), "return_cycle_days"].dropna()

fig, ax = plt.subplots(figsize=(8,5))
sns.histplot(return_cycle_data, bins=20, kde=True, color="steelblue", ax=ax)
ax.set_title("Refund Cycle Time Distribution")
ax.set_xlabel("Days from Delivery to Return Request")
st.pyplot(fig)

# --- Top Return Reasons ---
st.subheader("Top Return Reasons")
top_reasons = df["return_reason"].value_counts().reset_index()
top_reasons.columns = ["reason", "count"]

fig, ax = plt.subplots(figsize=(8,5))
sns.barplot(x="reason", y="count", data=top_reasons, palette="magma", ax=ax)
ax.set_title("Top Return Reasons")
ax.set_ylabel("Number of Returns")
ax.set_xlabel("Reason")
plt.xticks(rotation=30)
st.pyplot(fig)

# --- Segmentation: Region ---
st.subheader("Return Rate by Region")
region_returns = df.groupby("region")["is_return"].mean().reset_index()

fig, ax = plt.subplots(figsize=(8,5))
sns.barplot(x="region", y="is_return", data=region_returns, palette="coolwarm", ax=ax)
ax.set_title("Return Rate by Region")
ax.set_ylabel("Return Rate (%)")
st.pyplot(fig)
```
